# Feature Engineering

세션 메타데이터, history, current prompt에서 사용할 피처를 설계하고 검증하기 위한 노트북입니다.

## 현재 상태

아직 피처 실험은 진행하지 않았습니다. baseline 이후 후보 피처를 검증하면서 이 노트북을 채웁니다.

## 할 일

- prompt text 기반 피처 생성
- history action sequence 피처 생성
- session metadata 피처 생성
- workspace metadata 피처 생성
- 피처별 validation score 변화 기록
- 재사용 가능한 로직을 `../src/features/`로 이동
