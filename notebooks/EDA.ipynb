# EDA

DACON AI Agent 행동 예측 데이터의 구조와 분포를 확인하기 위한 노트북입니다.

## 현재 상태

아직 분석은 진행하지 않았습니다. 이 노트북은 앞으로 EDA를 시작할 때 사용할 기본 틀입니다.

## 할 일

- `train.jsonl`과 `train_labels.csv` 로드
- `id` 기준 정합성 확인
- action class 분포 확인
- `current_prompt` 길이 분석
- `history` 길이 분석
- session metadata별 action 분포 확인
- 주요 그래프를 `../results/figures/`에 저장
