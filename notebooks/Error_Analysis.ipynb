# Error Analysis

모델 예측 결과가 나온 뒤 오답 패턴을 분석하기 위한 노트북입니다.

## 현재 상태

아직 모델 학습과 예측을 진행하지 않았기 때문에 분석 결과는 없습니다. 첫 validation prediction이 생성된 뒤 이 노트북을 채웁니다.

## 할 일

- validation prediction 로드
- confusion matrix 생성
- class별 false positive / false negative 확인
- high-confidence 오답 검토
- 오류 유형을 `../docs/06_Error_Analysis.md`에 정리
